```python
import os
import pandas as pd
from astropy.io import fits
```

```python
# Set folder to look for fit files
folder = "/Users/danielbhuglah/HOPS Practice/HAT-P-16 b/hatp16b_20171012_holomon_cgepro_c11xlt_atik11000_red"

# Choose the FITS header keywords you want to extract
tags = ["DATE-OBS","TIME-OBS","EXPTIME", "AIRMASS", "FILTER","XBINNING","YBINNING","SET-TEMP","CCD-TEMP","XORGSUBF","YORGSUBF"]
```

```python
def classify_frame(filename):
    name = filename.lower()

    if "bias" in name:
        return "bias"
    if "dark" in name:
        return "dark"
    if "flat" in name:
        return "flat"

    return "science"   # default for science frames
```

```python
rows = []

for filename in os.listdir(folder):   
    if filename.lower().endswith(".fit"):
        filepath = os.path.join(folder, filename)

        # Open FITS file and read header
        with fits.open(filepath) as hdul:
            hdr = hdul[0].header

            # Extract requested tags
            row = {"filename": filename}
            for tag in tags:
                row[tag] = hdr.get(tag, None)   # None if missing

            # Add category based on filename
            row["category"] = classify_frame(filename)
            
            rows.append(row)

# Convert to DataFrame
df = pd.DataFrame(rows)

print(df)
```
