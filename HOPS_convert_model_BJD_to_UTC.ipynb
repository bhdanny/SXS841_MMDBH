```python
from astropy.time import Time
from astropy.coordinates import SkyCoord, EarthLocation
import astropy.units as u
import pandas as pd
```

```python
# read csv
df = pd.read_csv("/Users/danielbhuglah/Library/CloudStorage/OneDrive-TheOpenUniversity/SXS841 OSO Observations/lightcurve.csv")
```

```python
# target star coordinates - TOI-1266 c
target = SkyCoord(
    ra="13h11m59.18s",
    dec="+65d50m01.3s",
    frame="icrs"
)

# observatory location - PIRATE
obs = EarthLocation(
    lat=28.29925*u.deg,
    lon=-16.51019444*u.deg,
    height=2500*u.m
)
```

```python
# BJD_TDB column
t_bjd = Time(df["1. mid-exposure time in BJD_TDB"].values,
             format='jd',
             scale='tdb')

# reverse barycentric light travel time
ltt = t_bjd.light_travel_time(target,
                              location=obs,
                              kind='barycentric')

# convert back to UTC
t_utc = (t_bjd - ltt).utc

# add human-readable date/time
df["UTC"] = t_utc.iso

print(df[["1. mid-exposure time in BJD_TDB", "UTC"]])
```
