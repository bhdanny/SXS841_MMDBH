```python
# Import modules
from pathlib import Path
import pandas as pd
from datetime import datetime

# hops_tools contain the routines needed to collate info from directories and parse the HOPS output files
from hops_tools import (
    collect_fitting_results,
    extract_all_logs
)
```

```python
# Specify where HOPS outfiles are stored 
#root = Path("/Users/danielbhuglah/Library/CloudStorage/OneDrive-TheOpenUniversity/SXS841 OSO Observations/2026_02_24_files_consolidated/")
root = Path("/Users/danielbhuglah/Library/CloudStorage/OneDrive-TheOpenUniversity/2026_05_15_TOI-1803b/")
# where consolidated data should be downloaded too
downloads = root
```

```python
# Run both extraction routines
df_fitting = pd.DataFrame(collect_fitting_results(root))
df_photometry = extract_all_logs(root)

# Build output file name
timestamp = datetime.now().strftime("%Y%m%d_%H%M")
output_path = downloads / f"hops_combined_export_TOI_1266c_{timestamp}.xlsx"

# Write both DataFrames into one Excel file
with pd.ExcelWriter(output_path) as writer:
    df_fitting.to_excel(writer, sheet_name="Fitting_Results", index=False)
    df_photometry.to_excel(writer, sheet_name="Photometry_Logs", index=False)

print(f"Saved combined export to: {output_path}")
```
